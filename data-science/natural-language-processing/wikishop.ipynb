Создадим функцию для визуализации категориальных признаков. Функция будет строить столбчатую и круговую диаграммы для признака.

```python
def visualize_categorical_columns(df, category_cols=None):
    """
    Строит столбчатые и круговые диаграммы для категориальных признаков датасета.

    Параметры:
    - df (pandas.DataFrame): Исходный датасет.
    """
    # Получаем список категориальных столбцов
    if category_cols is None or not list(category_cols):
        category_cols = df.select_dtypes(exclude=["number"]).columns

    # Для каждого категориального столбца строим диаграммы
    for col in category_cols:
        # Частотная таблица для столбца
        freq_table = df[col].value_counts().reset_index()
        freq_table.columns = ["category", "frequency"]

        # Столбчатая диаграмма (bar-chart)
        plt.figure(figsize=(8, 6))
        ax = sns.barplot(
            x="category", y="frequency", data=freq_table, palette="viridis"
        )
        ax.set_title(f"Столбчатая диаграмма для {col}")
        ax.set_xlabel("Категория")
        ax.set_ylabel("Частота")
        plt.show()

        # Круговая диаграмма (pie chart)
        plt.figure(figsize=(8, 6))
        plt.pie(
            freq_table["frequency"],
            labels=freq_table["category"],
            autopct="%1.1f%%",
            startangle=90,
        )
        plt.title(f"Круговая диаграмма для {col}")
        plt.axis("equal")
        plt.show()
```

## BERT

### Подготовка

Выберем случайных 3К записей, т.к. не хватает памяти и обработка занимает продолжительное время

```python
df_toxic_comments_bert = df_toxic_comments.sample(SAMPLE_SIZE, random_state=RANDOM_STATE)
df_toxic_comments_bert.head()
```

```python
# модель toxic-bert поддерживает только 512 токенов
# дополительные 2 токена (чтобы получить 512) -- это начало и окончание строки
MAX_TOKEN_LENGTH = 510

tokenizer = transformers.BertTokenizer.from_pretrained(PRETRAINED_MODEL_NAME)

# truncate text to specified tokens number
def truncate_text(text, max_length):
    """Обрезаем текст до первых max_length токенов"""
    tokens = tokenizer.encode(text, add_special_tokens=True)

    # Проверяем длину токенов
    if len(tokens) <= max_length:
        return text  # Если длина меньше или равна max_length, возвращаем исходный текст
    else:
        # Иначе берем первые max_length токенов и обратно декодируем их в строку
        truncated_tokens = tokens[:max_length]
        truncated_text = tokenizer.decode(truncated_tokens, clean_up_tokenization_spaces=True)

        return truncated_text.strip()  # Удаляем лишние пробелы

# Применяем функцию ко всему датасету
df_toxic_comments_bert["truncated_text"] = df_toxic_comments_bert["text"].apply(lambda x: truncate_text(x, MAX_TOKEN_LENGTH))

tokenized = df_toxic_comments_bert["truncated_text"].apply(
    lambda x: tokenizer.encode(x, add_special_tokens=True)
)
```

```python
tokenized[:5]
```

Дополним все строки до максимальной длины

```python
max_len = 0
for i in tokenized.values:
    if len(i) > max_len:
        max_len = len(i)

padded = np.array([i + [0]*(max_len - len(i)) for i in tokenized.values])

attention_mask = np.where(padded != 0, 1, 0)
```

```python
padded[:5]
```

```python
padded.shape
```

Получим конфигурацию и модель BERT

```python
config = transformers.BertConfig.from_pretrained(PRETRAINED_MODEL_NAME)
model = transformers.BertModel.from_pretrained(PRETRAINED_MODEL_NAME, config=config)
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

model.to(device)
```

Обработаем строки

```python
batch_size = 100
embeddings = []
for i in notebook.tqdm(range(padded.shape[0] // batch_size)):
    batch = torch.LongTensor(padded[batch_size * i : batch_size * (i + 1)])
    attention_mask_batch = torch.LongTensor(
        attention_mask[batch_size * i : batch_size * (i + 1)]
    )

    with torch.no_grad():
        batch_embeddings = model(batch, attention_mask=attention_mask_batch)

    embeddings.append(batch_embeddings[0][:, 0, :].numpy())
```

Получим признаки для обучения моделей

```python
features = np.concatenate(embeddings)
print(features[:5])
```

```python
X = pd.DataFrame(features)
X.head()
```

Выделим целевой признак

```python
y = df_toxic_comments_bert['toxic']
y.head()
```

Раздедим выборки на тренировочную и тестовую

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=TEST_SPLIT_SIZE, stratify=y, random_state=RANDOM_STATE)
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

Данные подготовлены. Приступим к обучению моделей и подбору гиперпараметров.

#### Logistic Regression

```python
# Настраиваем пространство поиска гиперпараметров для Optuna
param_distributions = {
    "C": optuna.distributions.FloatDistribution(1e-3, 1e3, log=True),
    "solver": optuna.distributions.CategoricalDistribution(
        ["newton-cg", "lbfgs", "liblinear"]
    ),
    "max_iter": optuna.distributions.IntDistribution(100, 1000),
}

# Настраиваем процесс оптимизации
search_cv_lr = OptunaSearchCV(
    n_trials=30,
    estimator=LogisticRegression(random_state=RANDOM_STATE),
    param_distributions=param_distributions,
    cv=5,
    scoring="f1",
    random_state=RANDOM_STATE,
    verbose=False,
)

# Запускаем процедуру поиска наилучших гиперпараметров
search_cv_lr.fit(X_train, y_train)

# Показываем лучшие найденные гиперпараметры
print("Лучшие гиперпараметры:", search_cv_lr.best_params_)

# Оцениваем качество предсказания лучшей моделью
f1_score_cv_lr = search_cv_lr.best_score_
print(f"F1 Score CV: {f1_score_cv_lr:.3f}")
```

#### Decision Tree

```python
# Настраиваем пространство поиска гиперпараметров для Optuna
param_distributions = {
    "max_depth": optuna.distributions.IntDistribution(2, 32),
    "min_samples_split": optuna.distributions.IntDistribution(2, 20),
    "criterion": optuna.distributions.CategoricalDistribution(
        ["gini", "entropy"]
    ),
}

# Настраиваем процесс оптимизации
search_cv_dt = OptunaSearchCV(
    n_trials=30,
    estimator=DecisionTreeClassifier(random_state=RANDOM_STATE),
    param_distributions=param_distributions,
    cv=5,
    scoring="f1",
    random_state=RANDOM_STATE,
    verbose=False,
)

# Запускаем процедуру поиска наилучших гиперпараметров
search_cv_dt.fit(X_train, y_train)

# Показываем лучшие найденные гиперпараметры
print("Лучшие гиперпараметры:", search_cv_dt.best_params_)

# Оцениваем качество предсказания лучшей моделью
f1_score_cv_dt = search_cv_dt.best_score_
print(f"F1 Score CV: {f1_score_cv_dt:.3f}")
```

#### Random Forest

```python
# Настраиваем пространство поиска гиперпараметров для Optuna
param_distributions = {
    "max_depth": optuna.distributions.IntDistribution(2, 32),
    "n_estimators": optuna.distributions.IntDistribution(10, 100),
    "min_samples_split": optuna.distributions.IntDistribution(2, 20),
}

# Настраиваем процесс оптимизации
search_cv_rf = OptunaSearchCV(
    n_trials=30,
    estimator=RandomForestClassifier(random_state=RANDOM_STATE),
    param_distributions=param_distributions,
    cv=5,
    scoring="f1",
    random_state=RANDOM_STATE,
    verbose=False,
)

# Запускаем процедуру поиска наилучших гиперпараметров
search_cv_rf.fit(X_train, y_train)

# Показываем лучшие найденные гиперпараметры
print("Лучшие гиперпараметры:", search_cv_rf.best_params_)

# Оцениваем качество предсказания лучшей моделью
f1_score_cv_rf = search_cv_rf.best_score_
print(f"F1 Score CV: {f1_score_cv_rf:.3f}")
```

```python
def dict_to_markdown_table(data, value_col='Метрика', key_col='Модель'):
    """
    Преобразует словарь в таблицу Markdown.
    
    Параметры:
    - data: словарь {ключ: значение}
    - value_col: название столбца для значений (по умолчанию 'Метрика')
    - key_col: название столбца для ключей (по умолчанию 'Модель')
    """
    # Заголовки
    table = f"| {key_col} | {value_col} |\n"
    table += f"|:-----|:-----|\n"

    # Добавляем строки для каждого элемента словаря
    for key, value in data.items():
        table += f"| {key} | {value} |\n"

    return table
```

Функции для лемматизации и очистки строк

```python
# test
nlp = spacy.load("en_core_web_sm")
doc = nlp("The cats are running.")
lemmas = [token.lemma_ for token in doc]
print(lemmas)

lemm_text = " ".join(lemmas)
print(lemm_text)
```

```python
# как стало
def lemmatize(text, nlp):
    doc = nlp(text)
    lemmas = [token.lemma_ for token in doc]
    lemm_text = " ".join(lemmas)

    return lemm_text
```

```python
def clear_text(text):
    # Удаляем любые специальные символы, цифры, символы перевода строки и табуляции
    cleaned_text = re.sub(r'[\W\d]+', ' ', text)

    # Убираем лишние пробелы
    cleaned_text = re.sub(r'\s+', ' ', cleaned_text).strip()

    return cleaned_text
```

Лемматизируем строки.

Размер выборки также ограничен, чтобы обработка заняла приемлемое количество времени

#### Logistic Regression

```python
# Создаем пайплайн, состоящий из двух этапов: векторизация и логистическая регрессия
pipeline_lr = Pipeline([
    ('vector', TfidfVectorizer(stop_words=stopwords)),
    ('classifier', LogisticRegression(random_state=RANDOM_STATE))
])

# Настраиваем пространство поиска гиперпараметров для Optuna
param_distributions = {
    "classifier__C": optuna.distributions.FloatDistribution(1e-3, 1e3, log=True),
    "classifier__solver": optuna.distributions.CategoricalDistribution(
        ["newton-cg", "lbfgs", "liblinear"]
    ),
    "classifier__max_iter": optuna.distributions.IntDistribution(100, 1000),
}

# Настраиваем процесс оптимизации
search_cv_lr = OptunaSearchCV(
    n_trials=30,
    estimator=pipeline_lr,
    # estimator=LogisticRegression(random_state=RANDOM_STATE),
    param_distributions=param_distributions,
    cv=5,
    scoring="f1",
    random_state=RANDOM_STATE,
    verbose=False,
)

# Запускаем процедуру поиска наилучших гиперпараметров
search_cv_lr.fit(X_train, y_train)

# Показываем лучшие найденные гиперпараметры
print("Лучшие гиперпараметры:", search_cv_lr.best_params_)

# Оцениваем качество предсказания лучшей моделью
f1_score_cv_lr = search_cv_lr.best_score_
print(f"F1 Score CV: {f1_score_cv_lr:.3f}")
```

#### Decision Tree

```python
pipeline_dt = Pipeline([
    ('vector', TfidfVectorizer(stop_words=stopwords)),
    ('classifier', DecisionTreeClassifier(random_state=RANDOM_STATE))
])

# Настраиваем пространство поиска гиперпараметров для Optuna
param_distributions = {
    "classifier__max_depth": optuna.distributions.IntDistribution(2, 32),
    "classifier__min_samples_split": optuna.distributions.IntDistribution(2, 20),
    "classifier__criterion": optuna.distributions.CategoricalDistribution(
        ["gini", "entropy"]
    ),
}

# Настраиваем процесс оптимизации
search_cv_dt = OptunaSearchCV(
    n_trials=30,
    estimator=pipeline_dt,
    # estimator=DecisionTreeClassifier(random_state=RANDOM_STATE),
    param_distributions=param_distributions,
    cv=5,
    scoring="f1",
    random_state=RANDOM_STATE,
    verbose=False,
)

# Запускаем процедуру поиска наилучших гиперпараметров
search_cv_dt.fit(X_train, y_train)

# Показываем лучшие найденные гиперпараметры
print("Лучшие гиперпараметры:", search_cv_dt.best_params_)

# Оцениваем качество предсказания лучшей моделью
f1_score_cv_dt = search_cv_dt.best_score_
print(f"F1 Score CV: {f1_score_cv_dt:.3f}")
```

#### Random Forest

```python
pipeline_rf = Pipeline([
    ('vector', TfidfVectorizer(stop_words=stopwords)),
    ('classifier', RandomForestClassifier(random_state=RANDOM_STATE))
])

# Настраиваем пространство поиска гиперпараметров для Optuna
param_distributions = {
    "classifier__max_depth": optuna.distributions.IntDistribution(2, 32),
    "classifier__n_estimators": optuna.distributions.IntDistribution(10, 100),
    "classifier__min_samples_split": optuna.distributions.IntDistribution(2, 20),
}

# Настраиваем процесс оптимизации
search_cv_rf = OptunaSearchCV(
    n_trials=30,
    estimator=pipeline_rf,
    # estimator=RandomForestClassifier(random_state=RANDOM_STATE),
    param_distributions=param_distributions,
    cv=5,
    scoring="f1",
    random_state=RANDOM_STATE,
    verbose=False,
)

# Запускаем процедуру поиска наилучших гиперпараметров
search_cv_rf.fit(X_train, y_train)

# Показываем лучшие найденные гиперпараметры
print("Лучшие гиперпараметры:", search_cv_rf.best_params_)

# Оцениваем качество предсказания лучшей моделью
f1_score_cv_rf = search_cv_rf.best_score_
print(f"F1 Score CV: {f1_score_cv_rf:.3f}")
```
